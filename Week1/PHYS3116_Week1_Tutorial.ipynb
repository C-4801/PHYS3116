````python
def convert_ms_to_kmh(velocity_in_ms):
    '''
    Converts velocity from m/s to km/h
    :param velocity_in_ms: 
    :return: velocity_in_hours
    '''
    return velocity_in_ms * (60 * 60) / 1e3 
    
# We can specify how precisely to print our variables in strings using f-strings
print(convert_ms_to_kmh(5))
print(f'{convert_ms_to_kmh(5):.2f} km/h')  # Fixed Point precision, i.e. to 2 decimal points
print(f'{convert_ms_to_kmh(5):.2e} km/h')  # Scientific Notation to 2 decimals points 
````

You are given the following date:
- Suns distance from centre of MW $R = (8.40 \pm 0.4) kpc$
- Tangential velocity of the Sun $v = (252 \pm 11) km/s$
- Suns age is estimated to be around 4.5 billion years
- $G = 6.67 \times 10^{-11} m^3 kg^{-1} s^{-2}$
- $M_\odot = 2.00 \times 10^{30} kg$
- 1 $pc$ is $3.26 ly$ or $3.09 \times 10^{16}m$

````python
# 3) a)
# Write your code here

R       = 8.40  #kpc
dR      = 0.4   #kpc
v       = 252   #km/s
dv      = 11    #km/s
t       = 4.5e3 #Myr
G       = 6.67e-11 #m^3 kg^-1 s^-2
M_s     = 2.00e30 #kg

def pc_2_m(pc):
    m = pc*3.09e16
    return m

def Orbit_Period(R,v): #R in kpc, v in km/s
    t_o = 2*np.pi*(pc_2_m(R))/v #period in s
    return t_o/(60*60*24*365.25*1e6) #output in M years

t_o = Orbit_Period(R,v)

def dt_o(R,dR,v,dv):
    Dt_o = abs((2*np.pi/v)*pc_2_m(dR)) + abs(-(2*np.pi*pc_2_m(R)/(v**2))*dv)
    return Dt_o/(60*60*24*365.25*1e6) #output in M years

dt_o = dt_o(R,dR,v,dv)

print(str(f'{t_o:.1f}') +" +- "+str(f'{dt_o:.1f}')+" Myr")
````

````python
# 3) b)
# Write your code here

N_o = t/t_o

def dN_o(t,t_o,dt_o):
    return abs(-t/(t_o**2)*dt_o)

dN_o = dN_o(t,t_o,dt_o)

print(str(f'{N_o:.1f}') +" +- "+str(f'{dN_o:.1f}')+" Orbits")
````

````python
# 3) c)
# Write your code here

#========================================================================
#       m v^2     G m M    |  |      m v^2 r^2  |  |      v^2 r 
# Fc = ------- = --------  |=>| M = ----------- |=>| M = -------
#         r        r^2     |  |        r G m    |  |        G
#========================================================================

M_MW = ((v*1e3)**2)*(pc_2_m(R)*1e3)/G #Everything in SI units -> result in kg (see dimensional analysis below)
# v (km/s)*10^3 -> v (m/s) |=>| v (m/s)^2 -> v^2 (m^2/s^2)
# R (kpc) pc_2_m -> R (km) |=>| R (km)*10^3 -> R(m)
# v^2 (m^2/s^2) * R(m) -> M_MW*G (m^3 s^-2)
# M_MW*G (m^3 s^-2) / G (m^3 s^-2 kg^-1) -> M_MW (kg)

M_MW_TSM = (M_MW/M_s)*1e-12 #convert to trillion solar masses

#Should have done this at the start of this code but im starting the function by converting everything to SI. better late than never i guess :/
def dM_MW(v,dv,R,dR,G):
    v_SI = v*1e3
    dv_SI = dv*1e3
    R_SI = pc_2_m(R)*1e3
    dR_SI = pc_2_m(dR)*1e3

    dM_MW = abs((2*v_SI*R_SI/G)*dv_SI) + abs(((v_SI**2)/G)*dR_SI)
    return dM_MW 

dM_MW = dM_MW(v,dv,R,dR,G)
dM_MW_TSM = (dM_MW/M_s)*1e-12 #convert to trillion solar masses

print(str(f'{M_MW_TSM:.3f}') +" +- "+str(f'{dM_MW_TSM:.3f}')+" Trillion Solar Masses")
print("---")
M_MW_TSM_expected = 1.2
print("The mass of the milky way calculated using kepler's 3rd law is approximately one-"+ str(round(M_MW_TSM_expected/M_MW_TSM))+ "th of the accepted value.")
print("---")
print("This is due to the fact that only matter inside the radius of the sun's orbit around the MW, hence we miss all the mass contributed by...")
print("stars/gas/dark-matter that is further out than the sun around the MW.")
````

Q3)D:

Using the mass of visible matter to calculate the centripetal acceleration acting on the sun around the milky way (then using that to find the velocity of the sun around the milky way using circular motion eqn) misses the mass contributed by dark matter. At a radius of ~8kpc from the center of the milky way, baryonic matter makes up just over 50% of the total mass enclosed within a sphere of that radius (from looking at the plot on slide 26 of wk1 thurs lecture slides). This missed mass means we undershoot the centripetal acceleration felt by the sun and come up with a much slower velocity for the sun around the milky way.
